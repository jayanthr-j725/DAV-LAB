```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv("IRIS.csv")

print("Original Dataset:")
print(df.head())
```

```python
print("Missing values in each column:")
print(df.isnull())

print("\nNumber of missing values in each column:")
print(df.isnull().sum())
```

```python
mean_df = df.copy()
mean_df.loc[2, "sepal_length"] = np.nan
mean_df.loc[5, "sepal_length"] = np.nan
mean_df.loc[10, "sepal_width"] = np.nan

print("Before Mean Imputation:")
print(mean_df.loc[0:12])
```

```python


mean_df["sepal_length"] = mean_df["sepal_length"].fillna(
    mean_df["sepal_length"].mean()
)

mean_df["sepal_width"] = mean_df["sepal_width"].fillna(
    mean_df["sepal_width"].mean()
)

print("After Mean Imputation:")
print(mean_df.loc[0:12])
```

```python
print("Mean Sepal Length:", mean_df["sepal_length"].mean())
print("Mean Sepal Width:", mean_df["sepal_width"].mean())
```

```python


data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena', 'Arjun'],
    'Age': [21, 22, None, 23, 21, None],
    'Salary': [25000, None, 30000, 28000, None, 32000],
    'Department': ['CSE', 'ECE', 'CSE', None, 'ISE', 'ECE'],
    'City': ['Mysore', 'Bangalore', 'Mysore', 'Hubli', 'Bangalore', 'Mysore']
}

df_median = pd.DataFrame(data)

print(df_median)
```

```python
#
df_median['Salary'] = df_median['Salary'].fillna(
    df_median['Salary'].median()
)

print(df_median)
```

```python
print("Median Salary:", df_median['Salary'].median())
```

```python
data = {
    'Student': ['Anu', 'Ravi', 'Priya', 'Kiran', 'Meena'],
    'Department': ['CSE', 'ECE', None, 'ISE', None]
}

df_ffill = pd.DataFrame(data)

print("Before Forward Fill:")
print(df_ffill)
```

```python

df_ffill['Department'] = df_ffill['Department'].ffill()

print("\nAfter Forward Fill:")
print(df_ffill)
```

```python
from sklearn.preprocessing import LabelEncoder

data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}

df_label = pd.DataFrame(data)

print(df_label)
```

```python

encoder = LabelEncoder()

df_label['Department_Encoded'] = encoder.fit_transform(
    df_label['Department']
)

print(df_label)
```

```python

print(encoder.classes_)
```

```python
import pandas as pd

data = {
    'Name': ['Anu', 'Ravi', 'Priya', 'Kiran'],
    'Department': ['CSE', 'ECE', 'ISE', 'CSE']
}

df_onehot = pd.DataFrame(data)

print(df_onehot)
```

```python

encoded_df = pd.get_dummies(
    df_onehot,
    columns=['Department']
)

print(encoded_df)
```

```python


encoded_df = pd.get_dummies(
    df_onehot,
    columns=['Department'],
    dtype=int
)

print(encoded_df)
```
